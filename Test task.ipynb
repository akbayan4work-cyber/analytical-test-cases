```python
# 8 вопрос . Экперименты. 

import pandas as pd
from scipy import stats

file_path = 'Данные для тестового задания.xlsx'
df = pd.read_excel(file_path, sheet_name='Данные АБ тестов')
experiments = sorted(df['experiment_num'].unique())
print("РЕЗУЛЬТАТЫ СТАТИСТИЧЕСКОГО АНАЛИЗА А/Б ТЕСТОВ \n")

for exp in experiments:
    exp_data = df[df['experiment_num'] == exp]
    
    control_group = exp_data[exp_data['experiment_group'] == 'control']['revenue']
    test_group = exp_data[exp_data['experiment_group'] == 'test']['revenue']
    
    arpu_control = control_group.mean()
    arpu_test = test_group.mean()
    
    t_stat, p_value = stats.ttest_ind(test_group, control_group, equal_var=False)
    
    print(f"Эксперимент №{exp}:")
    print(f"  Количество пользователей: Контроль = {len(control_group)}, Тест = {len(test_group)}")
    print(f"  Средний чек (ARPU): Контроль = {arpu_control:.1f}, Тест = {arpu_test:.1f}")
    print(f"  Полученное значение p-value = {p_value:.3f}")
    
    if p_value < 0.05:
        if arpu_test > arpu_control:
            print("  Вывод: Разница значима. Тестовый вариант официально ЛУЧШЕ.")
        else:
            print("  Вывод: Разница значима. Тестовый вариант официально ХУЖЕ.")
    else:
        print("  Вывод: Разница НЕ значима. Изменения ни на что не повлияли.")
```

```python
# 18 вопрос.вывод по эксперименту

from statsmodels.stats.proportion import proportions_ztest

payments = [1003, 1099]       
visitors = [100047501, 100001055] 

z_stat, p_value = proportions_ztest(payments, visitors)

print(f"Конверсия А: {payments[0] / visitors[0] * 100:.5f}%")
print(f"Конверсия B: {payments[1] / visitors[1] * 100:.5f}%")
print(f"p-value: {p_value:.3f}")
if p_value < 0.05:
    print("\nВывод: разница статистически значима (p < 0.05)")
    print("Рекомендация: внедрить вариант B")
else:
    print("\nВывод: разница НЕ значима (p >= 0.05)")
    print("Рекомендация: не внедрять, оставить вариант A")
```
